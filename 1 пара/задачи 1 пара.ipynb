### Задача 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: 

>а) различные призы; 

>б) одинаковые призы?

* а. Различные призы(размещения с повторениями) $\tilde{A}_{10}^5 = 10^5$.


```python
n_films = 10
n_nominations = 5

answer_a = n_films ** n_nominations
print(f"{n_films}^{n_nominations} = {answer_a}")
```

* б. Одинаковые призы(сочетания с повторениями) $\tilde{C}_{10}^5 = C_{14}^5$.

```python
answer_b = comb(n_films + n_nominations - 1, n_nominations)
print(f"   C({n_films + n_nominations - 1}, {n_nominations}) = {answer_b}")
```

### Задача 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 — по 2 раза?

* Перестановки с повторениями $P_7(3; 2; 2) = \frac{7!}{3!2!2!}$

```python
total_digits = 7
count_4 = 3
count_5 = 2
count_6 = 2

answer = factorial(total_digits) // (factorial(count_4) * factorial(count_5) * factorial(count_6))

print(f"P({total_digits}; {count_4}, {count_5}, {count_6}) = {total_digits}! / ({count_4}! · {count_5}! · {count_6}!)")
print(f"= {answer}")
```

### Задача 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к другой: а) 3 карточки; б) все 6 карточек. 

>Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

```python
from math import perm, factorial  

letters = ['Т', 'Е', 'И', 'Я', 'Р', 'О']
n_letters = len(letters)

# a) Слово "ТОР" из 3 карточек
print(f"\na) Вероятность слова 'ТОР' (3 карточки):")

# Всего способов выбрать 3 карточки из 6 с учетом порядка
total_a = perm(n_letters, 3)  # A_6^3 = 6!/(6-3)!
print(f"   Всего размещений: A_{n_letters}^3 = {total_a}")

# Благоприятный исход (последовательность ТОР)
prob_a = 1 / total_a
print(f"   P(ТОР) = 1/{total_a} = {prob_a:.6f}")
```

* б) $P(B) = \frac{m}{n} = \frac{1}{P_6}$

```python
print(f"\nб) Вероятность слова 'ТЕОРИЯ' (6 карточек):")

# Всего перестановок 6 букв
total_b = factorial(n_letters)
print(f"   Всего перестановок: {n_letters}! = {total_b}")

# Благоприятный исход
prob_b = 1 / total_b
print(f"   P(ТЕОРИЯ) = 1/{total_b} = {prob_b:.6f}")
```

### Задача 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: 

>а) никакие два лица одного пола не сядут рядом; 

>б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

* а) $P(A) = \frac{m_1}{n} = \frac{2(5!)^2}{10!}$

```python
n_men = 5
n_women = 5
n_people = n_men + n_women

# а) Никто из одного пола не сидит рядом (чередование)
print(f"\nа) Чередование мужчин и женщин:")

# Всего способов рассадить 10 человек за круглым столом: (n-1)! для круговых перестановок
total_ways = factorial(n_people - 1)
print(f"Всего способов рассадки: ({n_people}-1)! = {total_ways}") 

# Фиксируем мужчин (5! способов), женщины между ними (5! способов), умножаем на 2
favorable_a = 2 * factorial(n_men) * factorial(n_women)
print(f"   Благоприятные: 2 · {n_men}! · {n_women}! = {favorable_a}")

prob_a = favorable_a / total_ways
print(f"   P(чередование) = {favorable_a}/{total_ways} = {prob_a:.6f}")
```

* б) $P(B) = \frac{m_2}{n} = \frac{5!2^5}{10!}$

```python
print(f"\nб) Супруги сидят рядом:")
# Всего способов рассадить 10 человек за круглым столом: (n-1)! для круговых перестановок
total_ways = factorial(n_people - 1)
print(f"Всего способов рассадки: ({n_people}-1)! = {total_ways}")

# Внутри каждой пары 2 перестановки: 2^5
favorable_b = factorial(n_men) * 2**n_men
print(f"   Благоприятные: {n_men}! · 2^{n_men} = {favorable_b}")

prob_b = favorable_b / total_ways
print(f"   P(супруги рядом) = {favorable_b}/{total_ways} = {prob_b:.6f}")
```

### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: 

>а) в два купе; 

>б) в семь купе; 

>в) в три купе.

* а) $P(A) = \frac{m_1}{n} = \frac{C_9^2 \cdot C_8^7}{C_{36}^7}$

```python
n_coupes = 9
seats_per_coupe = 4
n_passengers = 7
total_seats = n_coupes * seats_per_coupe

# Всего способов выбрать 7 мест из 36
total_outcomes = comb(total_seats, n_passengers)
print(f"\nВсего исходов: C({total_seats}, {n_passengers}) = {total_outcomes}")

# а) Все в 2 купе
print(f"\nа) 7 пассажиров в 2 купе:")

#Выбираем 2 купе из 9, в них 8 мест, нужно выбрать 7
#3+4 или 4+3 человека в купе
ways_choose_coupes = comb(n_coupes, 2)
ways_place_people = comb(seats_per_coupe, 3) * comb(seats_per_coupe, 4) + \
                    comb(seats_per_coupe, 4) * comb(seats_per_coupe, 3)

favorable_a = ways_choose_coupes * ways_place_people
prob_a = favorable_a / total_outcomes

print(f"Выбираем 2 купе: C({n_coupes}, 2) = {ways_choose_coupes}")
print(f"Размещаем 7 человек (3+4 или 4+3): {ways_place_people}")
print(f"Благоприятные: {favorable_a}")
print(f"P = {prob_a:.7f}")
```

* б) $P(B) = \frac{m_2}{n} = \frac{C_9^7 \cdot 4^7}{C_{36}^7}$

```python
print(f"\nб) 7 пассажиров в разных купе:")

# Выбираем 7 купе из 9
# В каждом купе - 1 место из 4
ways_choose_coupes_b = comb(n_coupes, n_passengers)
ways_choose_seats = seats_per_coupe ** n_passengers

favorable_b = ways_choose_coupes_b * ways_choose_seats
prob_b = favorable_b / total_outcomes

print(f"Выбираем 7 купе: C({n_coupes}, 7) = {ways_choose_coupes_b}")
print(f"В каждом 1 место из 4: {seats_per_coupe}^{n_passengers} = {ways_choose_seats}")
print(f"Благоприятные: {favorable_b}")
print(f"P = {prob_b:.6f}")
```

* в) $P(D) = \frac{m_3}{n} = \frac{C_9^3(6C_4^4C_4^2C_4^1 + 3C_4^3C_4^3C_4^1 + 3C_4^3C_4^2C_4^2)}{C_{36}^7}$

```python
print(f"\nв) 7 пассажиров в 3 купе:")

#7 человек на 3 купе:
# 1+2+4 (6 перестановок)
# 1+3+3 (3 перестановки)
# 2+2+3 (3 перестановки)

# 1+2+4
term1 = 6 * comb(seats_per_coupe, 1) * comb(seats_per_coupe, 2) * comb(seats_per_coupe, 4)
# 1+3+3
term2 = 3 * comb(seats_per_coupe, 1) * comb(seats_per_coupe, 3) * comb(seats_per_coupe, 3)
# 2+2+3
term3 = 3 * comb(seats_per_coupe, 2) * comb(seats_per_coupe, 2) * comb(seats_per_coupe, 3)

ways_choose_coupes_c = comb(n_coupes, 3)
favorable_c = ways_choose_coupes_c * (term1 + term2 + term3)
prob_c = favorable_c / total_outcomes

print(f"Выбираем 3 купе: C({n_coupes}, 3) = {ways_choose_coupes_c}")
print(f"Варианты размещения:")
print(f"1+2+4: {term1}")
print(f"1+3+3: {term2}")
print(f"2+2+3: {term3}")
print(f"Всего: {term1 + term2 + term3}")
print(f"Благоприятные: {favorable_c}")
print(f"P = {prob_c:.6f}")
```
